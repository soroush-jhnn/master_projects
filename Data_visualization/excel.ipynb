```python
import pandas as pd

df = pd.read_csv('WDISeries.csv')
df['Main Topic'] = df['Topic'].fillna('Unknown').apply(lambda x: x.split(':')[0])
top_main_topics = df['Main Topic'].value_counts().head(7)
```

```python
import xlsxwriter

workbook = xlsxwriter.Workbook("top_main_topics_donut.xlsx")
worksheet = workbook.add_worksheet("Data")

# Write headers
worksheet.write("A1", "Main Topic")
worksheet.write("B1", "Count")

# Write data
for i, (topic, count) in enumerate(top_main_topics.items(), start=2):
    worksheet.write(f"A{i}", topic)
    worksheet.write(f"B{i}", count)

# Create donut chart
chart = workbook.add_chart({"type": "doughnut"})

chart.add_series({
    "name": "Top 7 Main Topics",
    "categories": "=Data!$A$2:$A$8",
    "values": "=Data!$B$2:$B$8",
    "data_labels": {
        "percentage": True,
        "leader_lines": True
    }
})

chart.set_title({"name": "Donut Chart: Top 7 Main Indicator Categories"})
chart.set_legend({"position": "right"})

# Insert chart into worksheet
worksheet.insert_chart("D2", chart, {"x_scale": 1.4, "y_scale": 1.4})

workbook.close()
```
